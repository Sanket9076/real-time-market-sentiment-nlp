# 01. Exploratory Data Analysis (EDA)
## Real-Time Market Sentiment & NLP Analysis Platform

This notebook performs comprehensive EDA on the ingested financial news dataset:
- Data Quality & Missing Value Auditing
- Sentiment Distribution & Categorical Counts
- Publisher/Source Coverage Analysis
- Article Volume Trends Over Time

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
from src.data_ingestion import DataIngestion
from src.data_cleaning import DataCleaner
from src.sentiment_analysis import SentimentAnalyzer
from src.analysis import AnalyticsEngine

# Set aesthetic style
sns.set_theme(style='darkgrid')
plt.rcParams['figure.figsize'] = (10, 6)
```

```python
# 1. Ingest Data
ingestion = DataIngestion(mode='demo')
raw_df = ingestion.fetch_data()

# 2. Clean Data
cleaner = DataCleaner()
clean_df, quality_report = cleaner.clean_dataset(raw_df)
print('Data Quality Report:')
print(quality_report)
```

```python
# 3. Run Sentiment Pipeline
analyzer = SentimentAnalyzer()
df = analyzer.analyze_dataframe(clean_df, text_column='title')
df.head()
```

```python
# 4. Sentiment Label Breakdown
sent_counts = df['sentiment_label'].value_counts()
plt.figure(figsize=(7, 5))
sns.barplot(x=sent_counts.index, y=sent_counts.values, palette=['#2ECC71', '#95A5A6', '#E74C3C'])
plt.title('Global Sentiment Distribution Count')
plt.ylabel('Article Count')
plt.xlabel('Sentiment Class')
plt.savefig('../outputs/figures/eda_sentiment_distribution.png', bbox_inches='tight')
plt.show()
```

```python
# 5. Asset Level Coverage & Sentiment Boxplot
plt.figure(figsize=(12, 6))
sns.boxplot(data=df, x='asset', y='compound_score', palette='Set2')
plt.axhline(0, color='red', linestyle='--', alpha=0.5)
plt.title('Compound Score Dispersion per Financial Asset')
plt.savefig('../outputs/figures/eda_asset_sentiment_dispersion.png', bbox_inches='tight')
plt.show()
```
