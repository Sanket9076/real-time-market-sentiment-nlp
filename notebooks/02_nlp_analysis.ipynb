# 02. Natural Language Processing & Sentiment Modeling
## Real-Time Market Sentiment & NLP Analysis Platform

This notebook evaluates text preprocessing, TF-IDF n-gram extraction, baseline VADER sentiment scoring, and Supervised ML Classifier training.

```python
import pandas as pd
from src.data_ingestion import DataIngestion
from src.text_preprocessing import TextPreprocessor
from src.sentiment_analysis import SentimentAnalyzer
from src.ml_sentiment import MLSentimentClassifier
from src.analysis import AnalyticsEngine
```

```python
# Ingest & Preprocess
raw_df = DataIngestion(mode='demo').fetch_data()
preprocessor = TextPreprocessor()
df = preprocessor.process_dataframe(raw_df, text_column='title')
df = SentimentAnalyzer().analyze_dataframe(df, text_column='title')
df[['title', 'processed_text', 'sentiment_label', 'compound_score']].head()
```

```python
# Top Positive & Negative Keywords Extraction
analytics = AnalyticsEngine()
keywords = analytics.extract_keywords_by_sentiment(df, top_k=10)
print('Top Positive Keywords:', keywords['positive'])
print('Top Negative Keywords:', keywords['negative'])
```

```python
# Train Supervised ML Classifier (TF-IDF + Logistic Regression)
ml_classifier = MLSentimentClassifier()
eval_metrics = ml_classifier.train_and_evaluate(df, text_col='processed_text', label_col='sentiment_label')
print('ML Model Evaluation Metrics:')
for k, v in eval_metrics.items():
    print(f' - {k}: {v}')
```
